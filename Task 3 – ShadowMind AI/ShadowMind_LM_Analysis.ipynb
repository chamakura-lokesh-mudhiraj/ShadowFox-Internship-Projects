## Theory

**Natural Language Processing:** NLP is the field of AI that enables computers to process and analyze human language.

**Language Model:** A language model learns statistical and semantic patterns from text. It can classify, generate, summarize, translate, or answer questions depending on its training objective.

**Transformer:** Transformers use attention mechanisms to identify which words are important in context. Tokenization converts text into numerical tokens before the model processes it.

**LangChain:** LangChain is not the model itself. It is middleware used to organize prompts, chains, memory, tools, retrieval, and model calls in LLM applications.

## What Happens Internally

1. The pretrained model and tokenizer are loaded.
2. Input text is tokenized into numerical IDs.
3. The Transformer processes token relationships using attention.
4. The classification head predicts a sentiment label.
5. The model returns a confidence score.

```python
results = add_manual_evaluation(run_experiments(classifier, EXPERIMENTS))
results[["category", "text", "predicted_label", "confidence", "response_time_sec", "manual_evaluation"]]
```

## Visualization of Results

```python
plt.figure(figsize=(10, 5))
plt.bar(results["category"], results["confidence"], color="#2f7d6d")
plt.xticks(rotation=35, ha="right")
plt.ylim(0, 1.05)
plt.title("Model Confidence by Experiment Type")
plt.ylabel("Confidence Score")
plt.tight_layout()
plt.show()
```

```python
label_counts = results["predicted_label"].value_counts()

plt.figure(figsize=(5, 5))
plt.pie(label_counts, labels=label_counts.index, autopct="%1.1f%%", colors=["#3b82f6", "#ef4444"])
plt.title("Predicted Sentiment Distribution")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(results["category"], results["response_time_sec"], marker="o", color="#7c3aed")
plt.xticks(rotation=35, ha="right")
plt.title("Response Time by Experiment Type")
plt.ylabel("Seconds")
plt.tight_layout()
plt.show()
```

## Optional LangChain Integration

This section demonstrates how the same HuggingFace model can be wrapped through LangChain. LangChain acts as orchestration middleware. The model still comes from HuggingFace.

```python
try:
    from langchain_huggingface import HuggingFacePipeline

    langchain_model = HuggingFacePipeline(pipeline=classifier)
    print("LangChain wrapper created successfully:", type(langchain_model).__name__)
except Exception as exc:
    print("LangChain section skipped. Install langchain-huggingface if needed.")
    print(type(exc).__name__, exc)
```

## Ethical Considerations

Language models can reflect bias from training data. They may also produce confident outputs when context is incomplete. In sensitive fields like healthcare, finance, and education, predictions should be reviewed by humans. Privacy must be protected by avoiding unnecessary collection or storage of personal text.

## Conclusion

The selected language model performs well on clear sentiment classification tasks, especially simple positive and negative statements. It is less reliable on sarcasm, ambiguity, and mixed emotional context. These findings show why language models must be evaluated critically instead of treated as always-correct systems.

Future work can include fine-tuning on a custom dataset, comparing multiple models, building a retrieval augmented generation pipeline, and testing larger instruction-tuned LLMs.
